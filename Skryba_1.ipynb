```python
big_code = """
def train_model(X, y, *, lr=0.1, epochs=50):
    \"\"\"Mini SGD for logistic regression (demo).\"\"\"
    import math
    w = [0.0] * len(X[0])
    b = 0.0

    for epoch in range(epochs):
        for i, x in enumerate(X):
            # logit
            z = sum(wj * xj for wj, xj in zip(w, x)) + b
            # sigmoid
            p = 1.0 / (1.0 + math.exp(-z))
            # gradient
            g = p - y[i]
            # update
            for j in range(len(w)):
                w[j] -= lr * g * x[j]
            b -= lr * g

    return w, b
"""

# Pisanie litera-po-literze:
scribe.typewriter(big_code, delay_ms=12, new_cell=True, execute_after=False)
```

```python
import sys, time, random
from IPython import get_ipython

def type_to_output_and_insert(code: str,
                              cps: float = 60,             # znaki/sekundę
                              newline_pause_ms: int = 120, # pauza po \n
                              jitter_ms: int = 40          # losowa „ludzka” nierówność
                             ):
    """Symuluje pisanie w OUTPUT, a na końcu wstawia gotowy kod do nowej komórki."""
    for ch in code:
        sys.stdout.write(ch)
        sys.stdout.flush()

        # tempo + drobna losowość
        base = 1.0 / max(cps, 1)
        extra = 0.0
        if ch == '\n':
            extra = newline_pause_ms / 1000.0
        if jitter_ms > 0:
            extra += random.uniform(0, jitter_ms/1000.0)
        time.sleep(base + extra)

    # po „przepisaniu” w OUTPUT – wstaw kod do kolejnej komórki
    ip = get_ipython()
    if ip is not None:
        ip.set_next_input(code, replace=False)
```

```python
from IPython.display import Javascript, display
import json

def type_code_in_new_cell(code: str, cps: float = 25.0):
    js = f"""
    (function() {{
        var code = {json.dumps(code)};
        var cps = {float(cps)};
        var delay = 1000.0 / cps;

        var nb = Jupyter.notebook;
        var cell = nb.insert_cell_below('code');
        nb.select_next();
        cell.set_text('');

        var i = 0;
        function typeNext() {{
            if (i >= code.length) {{
                cell.code_mirror.refresh();
                return;
            }}
            var editor = cell.code_mirror;
            editor.replaceRange(code[i], editor.getCursor());
            i++;
            setTimeout(typeNext, delay);
        }}
        typeNext();
    }})();"""

    display(Javascript(js))
```
