# Notebook Scribe – demo

```
from notebook_scribe import scribe
```

## Funkcja

```

scribe.function(
    name="add",
    args=["a", "b"],
    doc="Zwraca sumę a+b.",
    body="return a + b",
    insert_cell=True
)
```

## Formuła

```
scribe.formula("y = 3x^2 + 2x + 1", insert_cell=True)
```

## Snippet

```
scribe.snippet("read_csv", path="data.csv", insert_cell=True)
```
